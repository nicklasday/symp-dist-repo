```python
g=Symp_symb(7)
E=g.ext_alg
C=g.cochain_complex
D=Prenorm_distr_Jacobi_7(g)
P=Geom_Prolongation(D)
```

```python
def check_equivar(c):
    p_inv=-exp(-2*h)*y*Y-h*H-e*E
    c_0=c.subs({y:0,h:0,e:0})
    r=p_inv.Ad(c_0,mod='CE')-c
    simplify_cochain(r)
    if r==C.elt({}): return True
    else: return r
```
