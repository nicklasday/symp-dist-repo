```python
g=Symp_symb(7)
E=g.ext_alg
C=g.cochain_complex
D=Free_distr_Jacobi_7(g)
P=Geom_Prolongation(D)
```

```python
y,h,e=symbols('y,h,e')
yn,hn,en=symbols('yn,hn,en')
K=IndexedBase('K')
Y,H,E,X,e1,e2,e3,e4,e5,e6,N=g.basis
p1_subs_dict={e:e+en,h:h+hn,y:exp(2*hn)*y+yn}
p1_inv=-exp(-2*hn)*yn*Y-hn*H-en*E
yn,hn,en=symbols('yn,hn,en')
p1_subs_dict={e:e+en,h:h+hn,y:exp(2*hn)*y+yn}
p1_inv=-exp(-2*hn)*yn*Y-hn*H-en*E
```

```python
def check_equivar(c):
    p_inv=-exp(-2*h)*y*Y-h*H-e*E
    c_0=c.subs({y:0,h:0,e:0})
    r=p_inv.Ad(c_0,mod='CE')-c
    simplify_cochain(r)
    if r==C.elt({}): return True
    else: return r
```

```python
F1,K1=P.normal_frame(1)
simplify_cochain(K1)
print('K1 =',K1)
```

```python
F2,K2=P.normal_frame(2)
simplify_cochain(K2)
print('K2 =',K2)
```

```python
F3,K3=P.normal_frame(3)
simplify_cochain(K3)
```

```python
MoreJS={}
for i in range(3,len(g.basis)):
    for j in range(3,len(g.basis)):
        for k in range(3,len(g.basis)):
            MoreJS[K[i,j,k,5]]=K[i,j,k,4,3]-K[i,j,k,3,4]-K[3,4,3]*K[i,j,k,3]-K[3,4,4]*K[i,j,k,4]

Norm_subs={}
for k in range(4,len(g.basis)-2):
    for k1 in range(3,k+1):
        Norm_subs[K[3,k,k1]]=0
        for i in range(3,len(g.basis)):
            Norm_subs[K[3,k,k1,i]]=0
            Norm_subs[K[3,k,k1,i]]=0
            for j in range(3,len(g.basis)):
                Norm_subs[K[3,k,k1,i,j]]=0

# Canonical section
Norm_subs[K[3,9,9]]=0
for k in range(3,len(g.basis)-1):
    Norm_subs[K[4,9,k]]=0
    Norm_subs[K[3,9,9,k]]=0
    for k1 in range(3, len(g.basis)-1):
        Norm_subs[K[4,9,k,k1]]=0
        Norm_subs[K[3,9,9,k,k1]]=0
```

```python
K3_final=K3.subs(MoreJS).subs(JS_dict).subs(Norm_subs)
simplify_cochain(K3_final)
```
