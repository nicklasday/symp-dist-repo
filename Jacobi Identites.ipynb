```python
g=Symp_symb(7)
E=g.ext_alg
C=g.cochain_complex
P=Geom_Prolongation(g,prenorm='none')
# Pi=Geom_Prolongation(g,prenorm='invol')
# Ps=Geom_Prolongation(g,prenorm='standard')
y,h,e=symbols('y,h,e')
yn,hn,en=symbols('yn,hn,en')
K=IndexedBase('K')
Y,H,E,X,e1,e2,e3,e4,e5,e6,N=g.basis
p1_subs_dict={e:e+en,h:h+hn,y:exp(2*hn)*y+yn}
p1_inv=-exp(-2*hn)*yn*Y-hn*H-en*E
```

```python
def check_equivar(c):
    r=c.subs(p1_subs_dict)-p1_inv.Ad(c)
    simplify_cochain(r)
    if r==C.elt({}): return True
    else: return r
```

```python
def total_back_substitute(d):
    """Back substitutes so that the values of d do not involve its keys
    INPUTS:
    * 'd' - a dictionary
    """
    effective=False
    complete_keys=set()
    while complete_keys!=set(d.keys()):
        for A in d:
            t=d[A].subs(d)
            if simplify(t)==simplify(d[A]): 
                complete_keys.add(A)
            else: 
                effective=True
                d[A]=t
    if effective: print('Back substitutions performed')
```

```python
def back_substitute(d,k,v):
    """Substitutes k-->v in the values of d
    INPUTS:
    * 'd' - a dictionary
    * 'k' - a symbol for substitution
    * 'v' - a value for substitution"""
    for A in d:
        d[A]=d[A].subs(k,v)
```

```python
def check_if_subs_are_needed(d):
    for A in d:
        if simplify(d[A].subs(d)-d[A])!=0: 
            print('Substitutions needed')
            return None
    print('No subsitutions needed')
```
