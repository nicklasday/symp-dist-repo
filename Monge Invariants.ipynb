```python
g=Symp_symb(7)
E=g.ext_alg
C=g.cochain_complex
P=Geom_Prolongation(g,prenorm='Monge')
# Pi=Geom_Prolongation(g,prenorm='invol')
# Ps=Geom_Prolongation(g,prenorm='standard')
y,h,e=symbols('y,h,e')
K=IndexedBase('K')
Y,H,E,X,e1,e2,e3,e4,e5,e6,N=g.basis
```

```python
yn,hn,en=symbols('yn,hn,en')
p1_subs_dict={e:e+en,h:h+hn,y:exp(2*hn)*y+yn}
p1_inv=-exp(-2*hn)*yn*Y-hn*H-en*E

def check_equivar(c):
    r=c.subs(p1_subs_dict)-p1_inv.Ad(c)
    simplify_cochain(r)
    if r==C.elt({}): return True
    else: return r
```

```python
def back_substitute(d,k,v):
    """Substitutes k-->v in the values of d
    INPUTS:
    * 'd' - a dictionary
    * 'k' - a symbol for substitution
    * 'v' - a value for substitution"""
    for A in d:
        d[A]=d[A].subs(k,v)
```

```python
def check_if_subs_are_needed(d):
    for A in d:
        if simplify(d[A].subs(d)-d[A])!=0: 
            print('Substitutions needed')
            return None
    print('No subsitutions needed')
```

## Jacobi Identities for n=6 through weight 2

```python
J_subs={}
```

#### Weight 1

```python
wght_1_list=[]
for i in range(3,len(g.basis)):
    for j in range(i+1,len(g.basis)):
        for k in range(j+1,len(g.basis)):
            for l in range(3,len(g.basis)):
                if g.basis[i].wght+g.basis[j].wght+g.basis[k].wght-g.basis[l].wght==-1:
                    if simplify(P.Jacobi_id(i,j,k,l))!=0: wght_1_list.append((i,j,k,l))
```

```python
K_list=[]
JI_ind_list=[A for A in wght_1_list]
for i in range(len(K_list)):
    s=solve(P.Jacobi_id(*JI_ind_list[i]).subs(J_subs),K_list[i])
    if len(s)>0: print('\n',JI_ind_list[i], 'has solutions for ',K_list[i])
    else: print(JI_ind_list[i], 'has no solutions for ',K_list[i])
    back_substitute(J_subs,K_list[i],s[0])
    J_subs[K_list[i]]=s[0]
```

```python
for A in wght_1_list:
    print('\n',A)
    print(P.Jacobi_id(*A))
```

```python
check_if_subs_are_needed(J_subs)
```

```python
for A in wght_1_list:
    if P.Jacobi_id(*A).subs(J_subs)!=0: print(A)
print('Jacobi Identity weight 1 test complete')
```

#### Weight 2

```python
wght_2_list=[]
for i in range(3,len(g.basis)):
    for j in range(i+1,len(g.basis)):
        for k in range(j+1,len(g.basis)):
            for l in range(3,len(g.basis)):
                if g.basis[i].wght+g.basis[j].wght+g.basis[k].wght-g.basis[l].wght==-2:
                    wght_2_list.append((i,j,k,l))
```

```python
# Add in derivatives
for A in list(J_subs.keys()):
    J_subs[P.abn_ind_der(A,3)]=P.abn_ind_der(J_subs[A],3)
    back_substitute(J_subs,P.abn_ind_der(A,3),P.abn_ind_der(J_subs[A],3))
    J_subs[P.abn_ind_der(A,4)]=P.abn_ind_der(J_subs[A],4)
    back_substitute(J_subs,P.abn_ind_der(A,4),P.abn_ind_der(J_subs[A],4))
```

```python
K_list=[K[3,5,3],K[4,6,5],K[4,7,6],K[4,8,7],K[4,9,8],K[5,10,10],K[5,7,7],K[5,8,8],K[5,9,9],K[6, 9, 10],
       K[6, 8, 9],K[7, 8, 10],K[4, 5, 3],0,K[4, 10, 9],K[6, 7, 8],0]
JI_ind_list=[A for A in wght_2_list if A not in []]
for i in range(len(K_list)):
    if K_list[i]!=0:
        t=P.Jacobi_id(*JI_ind_list[i]).subs(J_subs)
        s=solve(t,K_list[i])
        if len(s)>0: print(JI_ind_list[i],'has solutions for ',K_list[i])
        else: print('\n',JI_ind_list[i],'has no solutions for ',K_list[i])
        back_substitute(J_subs,K_list[i],s[0])
        J_subs[K_list[i]]=s[0]
```

```python
# Double check that substitutions are complete
check_if_subs_are_needed(J_subs)
```

```python
for A in wght_2_list:
    if simplify(P.Jacobi_id(*A).subs(J_subs))!=0: print(A)
print('Jacobi Identity weight 2 test complete')
```
