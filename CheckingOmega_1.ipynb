```python
def bracket(v1,v2): # In fundamental and horizontal frame (invariant horizontal lift)
    r=zeros(*shape(v1))
    w1,w2=[[0,0,0]+v[3:len(v)] for v in [v1,v2]]
    r+=Vector_Field(w1,D).bracket(Vector_Field(w2,D)).vec
    r+=v1[0]*diff(v2,y)+v1[1]*diff(v2,h)+v1[2]*diff(v2,e)
    r+=-(v2[0]*diff(v1,y)+v2[1]*diff(v1,h)+v2[2]*diff(v1,e))
    return r
```
