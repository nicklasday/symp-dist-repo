```python
def max_nz_ind(i,j):
    w=g.basis[i].wght+g.basis[j].wght
    if w>1: return -1
    if w<=g.basis[-1].wght: return len(g.basis)-1
    k=0
    while k<len(g.basis) and g.basis[k].wght>=w:
        # print('since g.basis',[k],'.wght =',g.basis[k].wght)
        k+=1
    return k-1

def QN_K(*ind):
    """Substitute with the relations of K_subs return zero 
    """
    i,j,k=ind[0:3]
    if i==j or k>max_nz_ind(i,j): return 0
    s=1
    temp=list(ind)
    if i>j: 
        s=-1
        temp[0]=ind[1]
        temp[1]=ind[0]
    if K[*temp] in P.K_subs: return s*P.K_subs[K[*temp]]
    return s*K[*temp]
```

```python
def Jacobi_Identity(i,j,k,l):
    r=QN_K(j,k,l,i)+QN_K(k,i,l,j)+QN_K(i,j,l,k)
    for a in range(3,len(g.basis)):
        r+=QN_K(i,a,l)*QN_K(j,k,a)
        r+=QN_K(j,a,l)*QN_K(k,i,a)
        r+=QN_K(k,a,l)*QN_K(i,j,a)
    return r
```

```python
JR={}
for i in range(3,len(g.basis)):
    for j in range(i,len(g.basis)):
        for k in range(3, len(g.basis)):
            for l in range(3, len(g.basis)):
                temp=Jacobi_Identity(i,j,k,l)
                if temp!=0:
                    JR[(i,j,k,l)]=temp
```
