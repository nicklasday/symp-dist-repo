# Computing Geometric Prolongations

## Mathematical Summary of Representations

### Frame Representations

Initially, we're given a frame $\omega:M\times\mathfrak m\to TM$ with curvature function $K_{-1}$, which induces a section $\sigma_0:M\to P^0$. The section induces a vertical coordinates $g_0\in \overline{G}^0$ on $P^0$ and a horizontal lift $\chi^00$, and consequently a frame $\Omega_0:P^0\times \mathfrak{g}^0\to TP^0$ according to 

$$
\Omega_0(\lambda_0,X) = \begin{cases}
        \zeta_{\lambda_0}\big(X\big)\quad \text{if}\ X\in \mathfrak{g}_0
        \\
        \chi^0\big(\Omega(-,g_0X)\big)(\lambda_0) \quad\text{if}\ X\in \mathfrak{m}
    \end{cases}
    \quad\text{where}\ \lambda_0 = R^0_g\sigma_0(x),
$$
which has curvature

$$
    K^{\Omega_0}\Big(R^0_{g_0}\big(\sigma_0(x)\big)\Big) = g_0.K^{\omega}(x)
$$

We'll normalize in weight 1 by choosing $f_1: M\to C^1_1(\mathfrak{m},\mathfrak{g})$ so that $\partial f_1\big(\sigma_0(x)\big) = -\text{proj}_\mathcal{N}^\perp\big(K^{\Omega_0}_1(\sigma_0(x))\big) = -\text{proj}_\mathcal{N}^\perp\big(K^{\omega}(x)\big)$. It can be shown that the frame $\omega_1=\Omega_0\circ g_0^{-1}\exp(f_1)g_0$ is normal in weight 1, and can therefore induces a section $\sigma_1: P^0\to P^1$; write $\widetilde \sigma_1 = \sigma_1\circ \sigma_0$. The section induces vertical coordinates $g_1\in \exp(\mathfrak{g}_1)$ on $P^1\to P^0$, and consequently horizontal lift $\chi^1$ and a frame $\Omega_1$ on $P^1$ according to

$$
\begin{gather*}
    \Omega_1(\lambda_1,X)
    =\begin{cases}
        \zeta^1_{\lambda_1}(X) & \text{if}\ X\in \overline{\mathfrak{g}}^1
        \\
        \chi^1\big(\omega_{1}(-,g_1X)\big)(\lambda_{1}) & \text{if}\ X\in \mathfrak m
    \end{cases}
\end{gather*}
$$

which has curvature
$$
    K^{\Omega_1}\Big(R^1_{g_1}\sigma_1\big(R^0_{g_0}\sigma_0(x)\big)\Big) = g_1.K^{\omega_1}\Big(R^0_{g_0}\sigma_0(x)\Big).
$$

In the program, the identity matrix will not represent $\Omega_1$, but rather

$$
\begin{gather*}
    \widetilde{\Omega}_1(\lambda_1,X)
    =\begin{cases}
        \zeta^1_{\lambda_1}(X) & \text{if}\ X\in \overline{\mathfrak{g}}^1
        \\
        \chi^1\big(\Omega_{0}(-,g_1X)\big)(\lambda_{1}) & \text{if}\ X\in \mathfrak m
    \end{cases}
\end{gather*}
$$

which has curvature

$$
    K^{\widetilde \Omega_1}\Big(R^1_{g_1}\sigma_1\big(R^0_{g_0}\sigma_0(x)\big)\Big) = g_1.K^{\Omega_0}\Big(R^0_{g_0}\sigma_0(x)\Big) = (g_0g_1).K^{\omega}(x).
$$

We can represent $\Omega_1$ in this frame as $\Omega_1(\lambda_1) = \widetilde \Omega_1(\lambda_1)\circ \big((g_0g_1)^{-1}e^{f_1}g_0g_1\big)$, or simply $\big((g_0g_1)^{-1}e^{f_1}g_0g_1\big)$.

### Vertical Coordinates on $P^1$

Canonical coordinates on adjoint group of $\overline{\mathfrak{g}}^1$ together with the section $\sigma_1:P^0\to P^1$ induced by $\omega_1$ give vertical coordinates on $P^1$. Namely, over a point $x\in M$, the coordinates
$$
    (y,h,e)\mapsto \sigma_1\big(\sigma_0(x)\big)\cdot \exp\Big(\text{ad}\big(h\cdot H + e\cdot E\big)\Big)\cdot\exp\Big(\text{ad}\big(y\cdot Y)\Big).
$$
Considering $(y,h,e)$ as coordinates on the adjoint group, we can see that for any $p = (y,h,e)$ in the adjoint group,
$$
 \zeta_p(Y) = \exp(e-h)\partial_y,
 \quad
 \zeta_p(H) = y\partial_y+\partial_h,
 \quad \text{and}\quad
 \zeta_p(E) = y\partial_y+\partial_e.
$$
These fundamental vector fields are the first three elements of the frames $\Omega_1$ and $\widetilde \Omega_1$.

### Normalization

At each stage, the frame $\omega_k$ with curvature $K^{\omega_{k}}$ normal up to weight $k$ will be given. As noted in "Frame Representations" above, the frame $\omega_1$ on $P^0$ is normal to weight 1 and its equivariant lift to $P^1$ can be represented by $\widetilde\Omega_1\cdot (g_1g_0)^{-1}e^{f_1}g_1g_0$, or just $(g_1g_0)^{-1}e^{f_1}g_1g_0$, where $f_1:P^0\to C^1_1(\mathfrak{m},\mathfrak{g})$ is chosen so that by $\partial f_1 = -\text{proj}_\mathcal{N}^\perp\big(K^{\omega}_1\big)$ along $y=h=e=0$.

Now for each $k>0$, we define $\omega_k$ as follows: Choose $f_k: M\to C^1_k(\mathfrak{m},\mathfrak{g})$ so that $\partial f_k\big(\widetilde\sigma_k(x)\big) = -\text{proj}_\mathcal{N}^\perp K^{\omega_{k-1}}_k\big(\widetilde\sigma_k(x)\big)$ along $y=h=e=0$. The new frame $\omega_k = \omega_{k-1}\cdot g^{-1}\exp(f_k)g$ has

$$
    K^{\omega_k}_k = K^{\omega_{k-1}}_{k} + \partial f_k = \text{proj}_{\mathcal{N}}K_k^{\omega_{k-1}}\quad\text{and}\quad K^{\omega_k}_{\leq k-1} = K^{\omega_{k-1}}_{\leq k-1}
$$

Notice that we only need to compute the curvatures only along $y=h=e=0$ for the purpose of normalization.

**Remark:** The computation of the curvature function can assume the component in $\text{Hom}\big(\overline{\mathfrak{g}}^1\wedge\mathfrak{g}\big)$ is zero

### Computing Curvature

We aim to compute $K_{k+1}^{\omega_k}$ along $\widetilde\sigma_1(x):=\sigma_1\sigma_0(x)$,given the frame $\omega_k$ (represented in $\widetilde \Omega_1$) and the lower curvature $K^{\omega_k}_{\leq k}$. We shall utilize the fact that $\omega_k = \widetilde \Omega_1(g_0g_1)^{-1}e^{f_1}e^{f_2}\cdots e^{f_k}(g_0g_1)$ is represented by a block unipotent upper triangular matrix, namely $(g_0g_1)^{-1}e^{f_1}e^{f_2}\cdots e^{f_k}(g_0g_1)$. In particular, for $X_i, X_j$ in the basis of $\mathfrak{m}$ of weights $i$ and $j$,

$$
    K_{k+1}^{\omega_k}(X_i,X_j) = \Big(\omega_k^{-1}\big([\omega_k(X_i),\omega_k(X_j)]\big) - K_{\leq k}^{\omega_k}(X_i,X_j) - [X_i,X_j]\Big)_{i+j+k+1}
    \\
    =
    \Big(\omega_k^{-1}\big\{[\omega_k(X_i),\omega_k(X_j)] - \omega_k\big(K_{\leq k}^{\omega_k}(X_i,X_j)+[X_i,X_j]\big)\big\}\Big)_{i+j+k+1}
$$

where the right hand side is in weights greater than or equal to $i+j+k+1$ before the projection to $\mathfrak{g}_{i+j+k+1}$, so the coordinatization in $\omega_k$ is nearly trivial. Further, we know the lower weight components of $K^{\omega_k}$, so the above can be computed as

$$
    K_{k+1}^{\omega_k}(X_i,X_j)
    = 
    \Big(\omega_k^{-1}\big\{[\omega_k(X_i),\omega_k(X_j)] - \omega_k\big(K_{\leq k}^{\omega_{k-1}}(X_i,X_j)+\partial f_k(X_i,X_j)+[X_i,X_j]\big)\big)\big\}\Big)_{i+j+k+1}
$$

This also allows us to only compute $\big([\omega_k(X_i),\omega_k(X_j)]\big)_{\geq i+j+k+1}$ since the lower weights in the right-hand side are zero. Observe however that the other terms may have components in weights greater than $i+j+k$, so we must use our previous computations.

### Lie Brackets on $P^1$

Let $X_i$ be a basis for $\mathfrak{g}$ and $Y_i^j\in C^\infty(P^1)$, so that $Y_i=Y_i^j\widetilde\Omega_1(X_j)\in \mathfrak{X}(P^1)$ is arbitrary. The method `bracket` computes the components of
$$
\widetilde\Omega_1^{-1}\big([Y_1,Y_2]\big) = \Big(Y_1\big(Y_2^k\big)-Y_2\big(Y_1^k\big)\Big)X_k + Y_1^iY_2^j\widetilde\Omega_1^{-1}[\widetilde\Omega_1(X_i),\widetilde\Omega_1(X_j)]
$$
in weight greater than or equal to `min_w`. The directional derivative components of this expression are computed in `dir_der`, while $Y_1^iY_2^j[\widetilde\Omega_1(X_i),\widetilde\Omega_1(X_j)]$ is computed in `Omega_1_bracket` (in weight greater than or equal to `min_w`)

### Computing `dir_der`

For any functions $f, Y^i\in C^\infty(P^1)$ we can write
$$
Y^i\widetilde\Omega_1(X_i)(f) = Y^0\cdot \zeta(Y)(f) + Y^1\cdot \zeta(H)(f) + Y^2\cdot \zeta(E)(f) + \sum_{i=3}^{2n-2}Y^i\cdot\widetilde\Omega_1(X_i)(f)
\\
= Y^0\cdot \big(\exp(e-h)\partial_y\big)(f) + Y^1\cdot \big(y\partial_y+\partial_h\big)(f) + Y^2\cdot \big(y\partial_y+\partial_e\big)(f) + \sum_{i=3}^{2n-2}Y^i\cdot\widetilde\Omega_1(X_i)(f)
$$

Let $A=\exp(eE+hH)\exp(yY)$. Since $\widetilde\Omega_1(X_i) = \chi^1\Big(\chi^0\big(A_i^j\cdot X_j\big)\Big)$ is horizontal (that is, tangent to the image of $\widetilde\sigma_1$), we can consider the vertical coordinates as formal variables and compute
$$
    \widetilde\Omega_1(X_i)(f) = A_i^j\cdot\omega(X_j)(f)
$$


```python
class Geom_Prolongation(object):
    def __init__(self,T,prenorm='invol'):
        """Constructs a geometric prolongation of the Symplectic symbol T given. If prenorm='invol',
        then the initial frame and K[i,j,k] will be assumed to satisfy the involutivity conditions from
        Doubrov's and Zelenko's 2009 London paper. If instead prenorm='standard', then the inital frame
        will be defined to have [X,e1]=e2,[X,e2]=e3, etc.
        """
        self.alg=T
        self.cochain_complex=T.cochain_complex
        # # The Adjoint matrix corresponding to a point p=(y,h,e)
        self.Ad_p=exp(symbols('h')*T.basis[1].ad_mat()+symbols('e')*T.basis[2].ad_mat())*exp(symbols('y')*T.basis[0].ad_mat())
        self.Ad_p_inv=self.Ad_p.inv()
        self.Omega_1_bb_cache={}
        self.K=IndexedBase('K')
        self.K_subs={}
        self.f=[None]
        self.df=[None]
        self.KN={}
        self.FN=[]
        self.harm_curv_cache=[0,self.cochain_complex.elt({})] # (max_weight, harm_curv)
        
        self.m=Integer((len(T.basis)-5)/2)
        
        # bracket relations (degree zero part of structure function)
        self.K_subs={}
        self.K_subs.update({self.K[3,a,a+1]:1 for a in range(4,2*self.m+3)}) # [X,ei]
        self.K_subs.update({self.K[3,2*self.m+3,2*self.m+4]:0}) # Last [X,ei]
        self.K_subs.update({self.K[a,2*self.m+7-a,2*self.m+4]:(-1)**(1+a) for a in range(4,self.m+4)}) # [ei, e_(2m+1-i)]
        self.K_subs.update({self.K[a,b,a+b-3]:0 for a in range(4,2*self.m+4) for b in range(4,2*self.m+7-a)}) #[ei,ej], i+j<2m+1
            
        # Involutivity Relations
        if prenorm=='invol':
            # First relation
            self.K_subs.update({self.K[a,b,c]:0 for b in range(4,self.m+4)
                                 for a in range(4,b) for c in range(b+1,a+b+1)})
            self.K_subs.update({self.K[a,b,3]:0 for b in range(4,self.m+4) for a in range(4,b)})
            # Second relation
            self.K_subs.update({self.K[a,b,c]:0 for b in range(self.m+4,2*self.m+4)
                                for a in range(4,2*self.m+7-b) for c in range(b+1,a+b-3)})
            
        if prenorm=='standard':
            # [X,e_i]=e_{i+1}
            for i in range(4,len(self.alg.basis)-2):
                self.K_subs.update({self.K[3,i,j]:0 for j in range(3,i+1)})
            #[e_1,e_{2n-6}]=N
            self.K_subs.update({self.K[4,len(self.alg.basis)-2,j]:0 for j in range(3, len(self.alg.basis)-1)})
            
        # Set the normal curvature
        for i in range(3,len(self.alg.basis)):
            for j in range(i,len(self.alg.basis)):
                self.KN[(i,j)]=zeros(len(self.alg.basis),1)
    
    def abn_ind_der(self,ind_expr,i):
        """Returns the derivative of ind_expr in the direction X_i among X_3, X_4,..., X_{2n-1},
        which is a frame on the base manifold.

        Note: This only works for horizontal derivatives!

        INPUTS:
        * 'ind_expr' -- an expression in h, e, y, and indexed objects
        * 'i' -- an integer between 3 and 2n-1
        """
        if type(ind_expr) in [Matrix, ImmutableDenseMatrix, MutableSparseMatrix, ImmutableSparseMatrix]:
            if type(ind_expr)==ImmutableDenseMatrix: result = mut_mat_copy(ind_expr)
            else: result = copy.copy(int_expr)
            for j in range(shape(result)[0]):
                for k in range(shape(result)[1]):
                    result[j,k]=self.abn_ind_der(result[j,k],i)
            return result

        # # I'm not sure if this simplification will help or hurt time efficiency
        # ind_expr=simplify(ind_expr)
        if isinstance(ind_expr,numbers.Number):
            return 0
        if type(ind_expr)==Add:
            result = Add(*[self.abn_ind_der(A,i) for A in ind_expr.args])
            return result
        if type(ind_expr)==Mul:
            result=0
            for j in range(len(ind_expr.args)):
                result+=self.abn_ind_der(ind_expr.args[j],i)*Mul(*list(ind_expr.args[0:j]+ind_expr.args[j+1:len(ind_expr.args)]))
            return result
        if type(ind_expr)==Pow:
            return ind_expr.exp*ind_expr.base**(ind_expr.exp-1)*self.abn_ind_der(ind_expr.base,i)
        if type(ind_expr)==Indexed:
            base=ind_expr.base
            ind=list(ind_expr.indices)
            return base[ind+[i]]
        if type(ind_expr)==Symbol: return 0
        if type(ind_expr)==exp:
            return ind_expr*self.abn_ind_der(ind_expr.args[0],i)
    
    def dir_der(self,f,v):
        """Returns the directional derivative of f in the direction of v
        INPUTS:
        * 'f' - an expression in h,e,y, and K
        * 'v' - a vector representation in Omega_1 of a vector field on P^1
        """
        y,h,e=symbols('y,h,e')
        dy=diff(f,y)
        dh=diff(f,h)
        de=diff(f,e)

        index_derivatives=Matrix([[self.abn_ind_der(f,i)] for i in range(3,len(self.alg.basis))])

        r=v[0]*exp(e-h)*dy+v[1]*(y*dy+dh)+v[2]*(y*dy+de)
        return r+(Matrix(v[3:len(v)]).transpose()*self.Ad_p[3:shape(self.Ad_p)[0],3:shape(self.Ad_p)[1]]*index_derivatives)[0]
    
    def Omega_1_base_bracket(self,i,j,min_w=-oo):
        """Returns the components of [Omega_1(X_i),Omega_1(X_j)] in weight >=min_w as a vector field 
        on P^1 represented in Omega_1. If i or j is less than 3, returns the zero vector field.

        INPUTS:
        * 'i','j' -- integers between 0 and len(self.alg.basis)
        """
        r=zeros(len(self.alg.basis),1)
        if i==j: return r
        if i>j: i1,j1,s=(j,i,-1)
        else: i1,j1,s=(i,j,1)
        
        if i<3 or j<3: return r
        
        B=self.alg.basis
        if (i1,j1) not in self.Omega_1_bb_cache:
            w=B[i1].wght+B[j1].wght
            for k in range(3,len(B)):
                if B[k].wght>=w:
                    r[k]+=self.K[i1,j1,k]
            self.Omega_1_bb_cache[(i1,j1)]=r.subs(self.K_subs)
            
        temp=zeros(len(self.alg.basis),1)+self.Omega_1_bb_cache[(i1,j1)]
        for i in range(len(self.alg.basis)):
            if self.alg.basis[i].wght<min_w: temp[i]=0
        return temp

    def Omega_1_bracket(self,Y1,Y2,min_w=-oo):
        """Freezes coefficients, then returns the components of [Y1,Y2] in weight >=min_w.

        INPUTS:
        * 'Y1','Y2' -- vector reps in Omega_1 of vector fields on P1; entries may involve h,e,y and K (which are considered constant)
        """
        r=zeros(len(self.alg.basis),1)
        for i in range(len(Y1)):
            for j in range(i+1,len(Y1)):
                r+=ind_diag(*[Y1[i]*Y2[j]-Y2[i]*Y1[j]]*len(self.alg.basis))*self.Omega_1_base_bracket(i,j,min_w)
        return r
    
    def bracket(self,Y1,Y2,min_w=-oo):
        """Returns the components of [Y1,Y2] of weight at least min_w for the vector fields Y1, Y2 on P^1

        INPUTS:
        * 'Y1','Y2' -- vector reps in Omega_1 of vector fields on P1; entries may involve h,e,y and K
        * 'min_w' -- an integer weight
        """
        r=zeros(len(self.alg.basis),1)
        for i in range(len(self.alg.basis)):
            if self.alg.basis[i].wght>=min_w:
                r[i]=[self.dir_der(Y2[i],Y1)-self.dir_der(Y1[i],Y2)]
        return r+self.Omega_1_bracket(Y1,Y2,min_w)
    
    def curv(self,F,l_curv,k):
        """Computes the curvature function for frame F on P^1 in weight precisely k.
        The curvature function is an element of C^2_+(m,g).
        INPUTS:
        * 'F', a block unipotent lower triangular matrix, representing in Omega_1 of a frame on P^1
        * 'l_curv' -- the curvature of F in wght < k as a cochain
        * 'k' -- natural number
        """
        # Convert lower curvature to dictionary form {(i,j): v such that F*v=[Fi,Fj]-[Xi,Xj]}
        lc=self.cochain_complex.curv_cochain_to_dict(l_curv)
        B=self.alg.basis

        if k<=0: return self.cochain_complex.elt({})
        
        r={}
        for i in range(3,len(B)):
            for j in range(i+1,len(B)):
                Z=self.bracket(F.col(i),F.col(j),B[i].wght+B[j].wght+k)-F*(lc[(i,j)]+B[i].ad_mat().col(j))
                # Z has entries in wghts => w(i)+w(j)+k
                s_ind = [a for a in range(len(B)) if B[a].wght==B[i].wght+B[j].wght+k]
                r[(i,j)]=zeros(len(self.alg.basis),1)
                for s in s_ind:
                    r[(i,j)][s]=Z[s]
        return self.cochain_complex.curv_dict_to_cochain(r)
    
    def normal_frame(self,k):
        """Returns a list [F,c], a frame on self which is normal up to degree k along with a cochain
        representing its curvature function
        INPUTS:
        * 'k' -- a natural number
        """
        C=self.cochain_complex
        if len(self.FN)>k: return (self.FN[k],self.KN)
        elif len(self.FN)<k:
            self.normal_frame(k-1)

        # Now set weight k assuming (k-1) is already done
        if k==0:
            self.KN=C.elt({})
            self.FN.append(eye(len(self.alg.basis)))
        else:
            time0=time.time()
            # Compute the wght k curvature of the previous frame
            pK=self.curv(self.FN[k-1],self.KN,k)
            time1=time.time()
            print('weight',k,'curvature computed in time',hrs_min_sec(time1-time0))

            # project onto im(d), choose fk for normalization
            # To Do: check if only computing curvatures along
            # y=h=e=0 would make this faster;
            # make sure it would remain correct
            dfk=-C.subspace_proj(pK,'exact')
            time2=time.time()
            print('weight',k,'projection computed in time',hrs_min_sec(time2-time1))

            
            fk=C.cb_preim_elt(dfk)
            time3=time.time()
            print('weight',k,'preimage cochain computed in time',hrs_min_sec(time3-time2))
            nK=pK+dfk

            # temporary
            self.f.append(fk)
            self.df.append(dfk)

            # modify the frame to nF
            step=int((self.alg.wght_list[0]-self.alg.wght_list[-1])/k+1)
            # Should I conjugate nilp_exp here?? I think not, but I was doing it before.
            # Indeed, I'm making f a function on P1 here, not on M
            nF=self.FN[k-1]*nilp_exp(simplify(C.one_cochain_mat_rep(fk)),step)
            time4=time.time()
            print('weight',k,'Matrix exponential complete in time', hrs_min_sec(time4-time3))

            self.KN=self.KN+nK
            self.FN.append(nF)
            print('weight',k,'normalization complete in time',hrs_min_sec(time.time()-time0))
        return self.normal_frame(k)
    
    def harm_curv(self,k):
        """The harmonic curvature of the canonical frame on self up to at least weight k. If higher weight
        curvatures have been computed, then the harmonic curvatures in those weights will be included.
        INPUTS:
        * 'k' -- a natural number
        """
        if self.harm_curv_cache[0]<k:
            self.harm_curv_cache[0]=k
            self.harm_curv_cache[1]=self.cochain_complex.subspace_proj(self.normal_frame(k)[1],'closed')
        return self.harm_curv_cache[1]
```

## Prenormalization

From the involutivity conditions $[V_i,V_i]\subseteq V_i$ and $[\mathcal{J}^{(i)},V_i]\subseteq \mathcal{J}^{(i)}$, we can impose the following conditions on the structure function $K_{ij}^k$:

$$ [\varepsilon_i,\varepsilon_j]=\sum_{k=1}^j K_{ij}^k\varepsilon_k,\quad \text{for}\ i<j\leq m$$
$$ [\varepsilon_i,\varepsilon_j]=K_{ij}^XX+\sum_{k=1}^j K_{ij}^k\varepsilon_k,\quad \text{for}\ j>m\ \text{and}\ i+j<2m+1$$
All this is captured in the below "prenormalization", but the indices are shifted to reflect the index of each element in the basis $(Y,H,E,X,\varepsilon_1,\ldots \varepsilon_6,\eta)$.
