## Mathematical Summary of Representations

### Frame Representations

The section induces a vertical coordinates $g_0\in \overline{G}^0$ on $P^0$ and a horizontal lift $\chi^0$, and consequently a frame $\Omega_0:P^0\times \mathfrak{g}^0\to TP^0$ according to 

$$
\Omega_0(\lambda_0,X) = \begin{cases}
        \zeta_{\lambda_0}\big(X\big)\quad \text{if}\ X\in \mathfrak{g}_0
        \\
        \chi^0\big(\Omega(-,g_0X)\big)(\lambda_0) \quad\text{if}\ X\in \mathfrak{m}
    \end{cases}
    \quad\text{where}\ \lambda_0 = R^0_g\sigma_0(x),
$$

which has curvature

$$
    K^{\Omega_0}\Big(R^0_{g_0}\big(\sigma_0(x)\big)\Big) = \text{Ad}(g_0^{-1})\big(K^{\omega}(x)\big)

We'll normalize in weight 1 by choosing $f_1: M\to C^1_1(\mathfrak{m},\mathfrak{g})$ so that $\partial f_1\big(\sigma_0(x)\big) = -\text{proj}_\mathcal{N}^\perp\big(K^{\Omega_0}_1(\sigma_0(x))\big) = -\text{proj}_\mathcal{N}^\perp\big(K^{\omega}(x)\big)$. It can be shown that the frame $\omega_1=\Omega_0\circ g_0^{-1}\exp(f_1)g_0$ is normal in weight 1, and can therefore induces a section $\sigma_1: P^0\to P^1;$ write $\widetilde \sigma_1 = \sigma_1\circ \sigma_0$. The section induces vertical coordinates $g_1\in \exp(\mathfrak{g}_1)$ on $P^1\to P^0$, and consequently horizontal lift $\chi^1$ and a frame $\Omega_1$ on $P^1$ according to

$$
\begin{gather*}
    \Omega_1(\lambda_1,X)
    =\begin{cases}
        \zeta^1_{\lambda_1}(X) & \text{if}\ X\in \overline{\mathfrak{g}}^1
        \\
        \chi^1\big(\omega_{1}(-,\pi_{\mathfrak{m}}g_1X)\big)(\lambda_{1}) -\zeta^1_{\lambda_1}(\pi_{\mathfrak{g}_+}g_1X)& \text{if}\ X\in \mathfrak m
    \end{cases}
\end{gather*}
$$

which has curvature
$$
    K^{\Omega_1}\Big(R^1_{g_1}\sigma_1\big(R^0_{g_0}\sigma_0(x)\big)\Big) = \text{Ad}(g_1^{-1})\Big(K^{\omega_1}\Big(R^0_{g_0}\sigma_0(x)\Big)\Big).
$$

In the program, the identity matrix will not represent $\Omega_1$, but rather

$$
\begin{gather*}
    \widetilde{\Omega}_1(\lambda_1,X)
    =\begin{cases}
        \zeta^1_{\lambda_1}(X) & \text{if}\ X\in \overline{\mathfrak{g}}^1
        \\
        \chi^1\big(\Omega_{0}(-,\pi_{\mathfrak{m}}g_1X)\big)(\lambda_{1})-\zeta^1_{\lambda_1}(\pi_{\mathfrak{g}_+}g_1X) & \text{if}\ X\in \mathfrak m
    \end{cases}
\end{gather*}
$$

which has curvature

$$
    K^{\widetilde \Omega_1}\Big(R^1_{g_1}\sigma_1\big(R^0_{g_0}\sigma_0(x)\big)\Big) = \text{Ad}(g_1^{-1})\Big(K^{\Omega_0}\Big(R^0_{g_0}\sigma_0(x)\Big)\Big)= \text{Ad}\big((g_0g_1)^{-1}\big)\big(K^{\omega}(x)\big).
$$

where $g_1:\mathfrak{m}\to \mathfrak{g}^0$ is projected. We can represent $\Omega_1$ in this frame as $\Omega_1(\lambda_1) = \widetilde \Omega_1(\lambda_1)\circ \big((g_0g_1)^{-1}e^{f_1}g_0g_1\big)$, or simply $\big((g_0g_1)^{-1}e^{f_1}g_0g_1\big)$.

### Vertical Coordinates on $P^1$

Canonical coordinates on adjoint group of $\overline{\mathfrak{g}}^1$ together with the section $\sigma_1:P^0\to P^1$ induced by $\omega_1$ give vertical coordinates on $P^1$. Namely, over a point $x\in M$, canonical coordinates of the second kind are given by
$$
    (y,h,e)\mapsto \sigma_1\big(\sigma_0(x)\big)\cdot \exp\Big(\text{ad}\big(h\cdot H + e\cdot E\big)\Big)\cdot\exp\Big(\text{ad}\big(y\cdot Y)\Big).
$$
Considering $(y,h,e)$ as coordinates on the adjoint group, we can see that for any $p = (y,h,e)$ in the adjoint group,
$$
 \zeta_p(Y) = \partial_y,
 \quad
 \zeta_p(H) = 2y\partial_y+\partial_h,
 \quad \text{and}\quad
 \zeta_p(E) = \partial_e.
$$
These fundamental vector fields are the first three elements of the frames $\Omega_1$ and $\widetilde \Omega_1$.

In these coordinates, the right action is given by 
$$
R_{(y_1,h_1,e_1)}(y_0,h_0,e_0) = (e^{2h_1}y_0+y_1,h_0+h_1,e_0+e_1),
$$
so inverses are given by 
$$
(y,h,e)^{-1} = (-y\cdot e^{-2h},-h,-e).
$$
Curvature functions $K$ which are equivariant should thus satisfy
$$
K(e^{2h_1}y_0+y_1,h_0+h_1,e_0+e_1) = \text{Ad}(-y_1e^{-2h_1},-h_1,-e_1)\big(K(y_0,h_0,e_0)\big)
$$
One can change to canonical coordinates of the first kind using the formula
$$
(y,h,e)= \exp(hH+eE)\exp(yY) = \exp\Big(hH+eE+\frac{2hy}{e^{2h}-1}Y\Big).
$$

### Normalization

At each stage, the frame $\omega_k$ with curvature $K^{\omega_{k}}$ normal up to weight $k$ will be given. As noted in "Frame Representations" above, the frame $\omega_1$ on $P^0$ is normal to weight 1 and its equivariant lift to $P^1$ can be represented by $\widetilde\Omega_1\cdot (g_1g_0)^{-1}e^{f_1}g_1g_0$, or just $(g_1g_0)^{-1}e^{f_1}g_1g_0$, where $f_1:P^0\to C^1_1(\mathfrak{m},\mathfrak{g})$ is chosen so that by $\partial f_1 = -\text{proj}_\mathcal{N}^\perp\big(K^{\omega}_1\big)$ along $y=h=e=0$.

Now for each $k>0$, we define $\omega_k$ as follows: Choose $f_k: M\to C^1_k(\mathfrak{m},\mathfrak{g})$ so that $\partial f_k\big(\widetilde\sigma_k(x)\big) = -\text{proj}_\mathcal{N}^\perp K^{\omega_{k-1}}_k\big(\widetilde\sigma_k(x)\big)$ along $y=h=e=0$. The new frame $\omega_k = \omega_{k-1}\cdot g^{-1}\exp(f_k)g$ has

$$
    K^{\omega_k}_k = K^{\omega_{k-1}}_{k} + \partial f_k = \text{proj}_{\mathcal{N}}K_k^{\omega_{k-1}}\quad\text{and}\quad K^{\omega_k}_{\leq k-1} = K^{\omega_{k-1}}_{\leq k-1}
$$

Notice that we only need to compute the curvatures only along $y=h=e=0$ for the purpose of normalization.

### Computing Curvature

We aim to compute $K_{k+1}^{\omega_k}$ along $\widetilde\sigma_1(x):=\sigma_1\sigma_0(x)$, given the frame $\omega_k$ (represented in $\widetilde \Omega_1$) and the lower curvature $K^{\omega_k}_{\leq k}$. We shall utilize the fact that $\omega_k = \widetilde \Omega_1(g_0g_1)^{-1}e^{f_1}e^{f_2}\cdots e^{f_k}(g_0g_1)$ is represented by a block unipotent upper triangular matrix, namely $(g_0g_1)^{-1}e^{f_1}e^{f_2}\cdots e^{f_k}(g_0g_1)$. In particular, for $X_i, X_j$ in the basis of $\mathfrak{m}$ of weights $i$ and $j$,

$$
    K_{k+1}^{\omega_k}(X_i,X_j) = \Big(\omega_k^{-1}\big([\omega_k(X_i),\omega_k(X_j)]\big) - K_{\leq k}^{\omega_k}(X_i,X_j) - [X_i,X_j]\Big)_{i+j+k+1}
    \\
    =
    \Big(\omega_k^{-1}\big\{[\omega_k(X_i),\omega_k(X_j)] - \omega_k\big(K_{\leq k}^{\omega_k}(X_i,X_j)+[X_i,X_j]\big)\big\}\Big)_{i+j+k+1}
$$

where the right hand side is in weights greater than or equal to $i+j+k+1$ before the projection to $\mathfrak{g}_{i+j+k+1}$, so the coordinatization in $\omega_k$ is nearly trivial. Further, we know the lower weight components of $K^{\omega_k}$, so the above can be computed as

$$
    K_{k+1}^{\omega_k}(X_i,X_j)
    = 
    \Big(\omega_k^{-1}\big\{[\omega_k(X_i),\omega_k(X_j)] - \omega_k\big(K_{\leq k}^{\omega_{k-1}}(X_i,X_j)+\partial f_k(X_i,X_j)+[X_i,X_j]\big)\big)\big\}\Big)_{i+j+k+1}
$$

This also allows us to only compute $\big([\omega_k(X_i),\omega_k(X_j)]\big)_{\geq i+j+k+1}$ since the lower weights in the right-hand side are zero. Observe however that the other terms may have components in weights greater than $i+j+k$, so we must use our previous computations.

### Lie Brackets on $P^1$

Let $X_i$ be a basis for $\mathfrak{g}$ and $Y_i^j\in C^\infty(P^1)$, so that $Y_i=Y_i^j\widetilde\Omega_1(X_j)\in \mathfrak{X}(P^1)$ is arbitrary. The method `bracket` computes the components of
$$
\widetilde\Omega_1^{-1}\big([Y_1,Y_2]\big) = \Big(Y_1\big(Y_2^k\big)-Y_2\big(Y_1^k\big)\Big)X_k + Y_1^iY_2^j\widetilde\Omega_1^{-1}[\widetilde\Omega_1(X_i),\widetilde\Omega_1(X_j)]
$$
in weight greater than or equal to `min_w`. The directional derivative components of this expression are computed in `dir_der`, while $Y_1^iY_2^j\widetilde\Omega_1^{-1}[\widetilde\Omega_1(X_i),\widetilde\Omega_1(X_j)]$ is computed in `Omega_1_bracket` (in weight greater than or equal to `min_w`)

### Computing `dir_der`

For any function $f\in C^\infty(P^1)$ we can write
$$
A^i_j\widetilde\Omega_1(X_i)(f) = Z_j
$$
where $Z_0=\partial_y, Z_1=\partial_h, Z_2=\partial_e, Z_3=\chi^1(-,X), Z_4=\chi^1(-,\varepsilon_1),$ etc. The matrix $A$ is called `cof_mat` and is computed at initialization of a `Geom_Prolongation`


```python
class Geom_Prolongation(object):
    def __init__(self,D,J_subs_dict):
        """Constructs a geometric prolongation of the Symplectic symbol T given.
        """

        self.alg=D.Tanaka_symbol
        self.distr=D
        self.jsd=J_subs_dict
        self.cochain_complex=self.alg.cochain_complex

        # # The Adjoint matrix corresponding to a point p=(y,h,e)
        self.Ad_p=exp(symbols('h')*self.alg.basis[1].ad_mat()+symbols('e')*self.alg.basis[2].ad_mat())*exp(symbols('y')*self.alg.basis[0].ad_mat())
        self.Ad_p_inv=self.Ad_p.inv()

        self.K=IndexedBase('K')
        self.gf=[None] # Temporary for testing
        self.gdf=[None] # Temporary for testing
        self.pK=[None]
        self.KN=self.cochain_complex.elt({})
        self.FN=eye(len(self.alg.basis))
        self.FN_inv=eye(len(self.alg.basis))
        self.harm_curv_cache=[0,self.cochain_complex.elt({})] # (max_weight, harm_curv)
        self.norm_wght=0

        self.m=Integer((len(self.alg.basis)-5)/2)
        self.yhe_pt=self.alg.elt([symbols('y'),symbols('h'),symbols('e')]+[0]*(len(self.alg.basis)-3))
        # This line sometimes takes a bit of time
        self.Omega_1_curv=self.alg.second_kind_inv(self.yhe_pt).Ad(self.distr.curv,mod='CE')
    
        # changes the frame from Omega_1 to dy, dh, de and horizontal VFs
        self.cof_mat=self.change_of_frame_mat()

    def change_of_frame_mat(self):
        # change the frame from Omega_1 to fundamental VFs and horizontal VFs
        cof1=eye(len(self.alg.basis))
        cof1[3:len(self.alg.basis),3:len(self.alg.basis)]=self.Ad_p[3:len(self.alg.basis),3:len(self.alg.basis)]
        temp1=zeros(len(self.alg.basis))
        temp1[0:3,0:len(self.alg.basis)]=self.Ad_p[0:3,0:len(self.alg.basis)]
        cof1[0:3,3:len(self.alg.basis)]=(self.Ad_p_inv*temp1)[0:3,3:len(self.alg.basis)]

        # change of frame matrix from fundamental and horizontal to dy, dh, de and horizontal
        cof2=eye(len(self.alg.basis))
        cof2[0,1]=2*symbols('y')
        return cof2*cof1
    
    def dir_der(self,f,v,normal=False):
        '''Returns the directional derivative of f in the direction of v
        INPUTS:
        * 'f' - an expression in h,e,y, and K
        * 'v' - a vector representation in Omega_1 of a vector field on P^1
        '''
        if type(f)==T_symb_elt:
            return f.parent.elt([self.dir_der(a,v) for a in f.vec])
        if type(f) in [Matrix, SparseMatrix]:
            r=zeros(*shape(f))
            for i in range(shape(f)[0]):
                for j in range(shape(f)[1]):
                    r[i,j]=self.dir_der(f[i,j],v)
            return r

        y,h,e=symbols('y,h,e')
        dy=diff(f,y)
        dh=diff(f,h) # It's slightly inefficient to compute these every time, but not a huge deal (probably)
        de=diff(f,e)

        der_vec=[dy,dh,de]+[0]*(len(self.alg.basis)-3)

        # self.cof_mat is change of frame from Omega_1 to vertical and horizontal derivatives
        cof_v=self.cof_mat*v
        for i in range(3,len(v)):
            if cof_v[i]!=0: 
                if normal: der_vec[i]=self.distr.normal_ind_der(f,i) # This was abn_ind_der before! Might have caused issues
                else: der_vec[i]=self.distr.abn_ind_der(f,i)
        der_vec=Matrix([der_vec])
        
        return (der_vec*cof_v)[0,0]
    
    def fund_der(self,f,*dir_list,F):
        r'''Returns the fundamental derivative of f in the directions of dir_list.
        
        args:
        * 'f' - an expression in h, e, y, and K
        * 'dir_list' - a list of indices which refer to the colums of F
        * 'F' - the matrix expression of a frame in \tilde\Omega_1
        '''
        r=copy.copy(f)
        for i in dir_list:
            r=self.dir_der(f,F.col(i))
        return r
    
    def Omega_1_bracket(self,Y1,Y2,min_w=-oo):
        """Freezes coefficients, then returns the components of [Y1,Y2] in weight >=min_w.

        INPUTS:
        * 'Y1','Y2' -- vector reps in Omega_1 of vector fields on P1; 
          entries may involve h,e,y and K (which are considered constant)
        """
        r=self.alg.elt(Y1).ad(self.alg.elt(Y2)).vec
        # Curvature must be added to r
        Z1=self.alg.elt([0,0,0]+Y1[3:len(Y1)])
        Z2=self.alg.elt([0,0,0]+Y2[3:len(Y2)])
        ext_elt=Z1.cast_as_ext_elt().wedge(Z2.cast_as_ext_elt())
        r=r+self.Omega_1_curv.apply_cochain_map(ext_elt).vec
        for i in range(len(r)):
            if self.alg.basis[i].wght<min_w: r[i]=0
        return r
    
    def bracket(self,Y1,Y2,min_w=-oo):
        """Returns the components of [Y1,Y2] of weight at least min_w for the vector fields Y1, Y2 on P^1

        INPUTS:
        * 'Y1','Y2' -- vector reps in Omega_1 of vector fields on P1; entries may involve h,e,y and K
        * 'min_w' -- an integer weight
        """
        r=zeros(len(self.alg.basis),1)
        for i in range(len(self.alg.basis)):
            if self.alg.basis[i].wght>=min_w:
                r[i]=[self.dir_der(Y2[i],Y1)-self.dir_der(Y1[i],Y2)]
        return r+self.Omega_1_bracket(Y1,Y2,min_w)
    
    def curv(self,F,l_curv,k):
        """Computes the curvature function for frame F on P^1 in weight precisely k.
        The curvature function is an element of C^2_+(m,g).
        INPUTS:
        * 'F', a block unipotent lower triangular matrix, representing in Omega_1 of a frame on P^1
        * 'l_curv' -- the curvature of F in wght < k as a cochain
        * 'k' -- natural number
        """
        B=self.alg.basis
        if k<=0: return self.cochain_complex.elt({})
        
        r={}
        for i in range(3,len(B)):
            for j in range(i+1,len(B)):
                Z=self.bracket(F.col(i),F.col(j),B[i].wght+B[j].wght+k)-F*(
                    l_curv.apply_cochain_map_base(B[i],B[j]).vec+B[i].ad_mat().col(j))
                # Z has entries in wght w(i)+w(j)+k
                s_ind = [a for a in range(len(B)) if B[a].wght==B[i].wght+B[j].wght+k]
                r[(i,j)]=zeros(len(self.alg.basis),1)
                for s in s_ind:
                    r[(i,j)][s]=Z[s]
        return self.cochain_complex.curv_dict_to_cochain(r)

    def normal_frame(self,k):
        """Returns a list [F,c], a frame on self which is normal up to degree k along with a cochain
        representing its curvature function up to degree k
        INPUTS:
        * 'k' -- a natural number
        """
        C=self.cochain_complex
        if k<=self.norm_wght: return (self.FN,self.KN)
        elif k>self.norm_wght+1: self.new_normal_frame(k-1,curv_v)

        # Now set weight k assuming (k-1) is already done
        time0=time.time()
        # Compute the wght k curvature of the previous frame
        self.pK.append(self.KN+self.curv(self.FN,self.KN,k))
        time1=time.time()
        print('weight',k,'curvature computed in time',hrs_min_sec(time1-time0))

        # project onto im(d) along the section, 
        hey_z={symbols('y'):0,symbols('h'):0,symbols('e'):0}
        dfk=-C.subspace_proj(self.pK[k].subs(hey_z),'exact') # along the section
        time2=time.time()
        print('weight',k,'projection computed in time',hrs_min_sec(time2-time1))

        # Choose fk from the preimage for normalization
        fk=C.cb_preim_elt(dfk) # along the section, for modifying the frame
        time3=time.time()
        print('weight',k,'preimage cochain computed in time',hrs_min_sec(time3-time2))
        gdfk=self.alg.second_kind_inv(self.yhe_pt).Ad(dfk,mod="CE")
        gfk=self.alg.second_kind_inv(self.yhe_pt).Ad(fk,mod="CE") # I think this will give a Cartan connection...
        self.KN=self.pK[k]+gdfk.wght_proj(k) # I think this should be "equivariant"...equivariant in the lowest nonzero weight
        # since wght 3 KN is not equivariant, one of the above is not equivariant
        
        # temporary for testing
        self.gf.append(gfk)
        self.gdf.append(gdfk)

        # modify the frame
        step=int((self.alg.wght_list[0]-self.alg.wght_list[-1])/k+1)
        gfk_equiv_mat=simplify(C.one_cochain_mat_rep(gfk))
        exp_gfk,exp_gfk_inv=nilp_exp(gfk_equiv_mat,step)
        self.FN=self.FN*exp_gfk
        self.FN_inv=exp_gfk_inv*self.FN_inv

        time4=time.time()
        print('weight',k,'Matrix exponential complete in time', hrs_min_sec(time4-time3))
        self.norm_wght=k
        
        self.KN=ds_subs(self.distr.normalize_der(self.KN),self.jsd,self.distr)
        self.FN=ds_subs(self.FN,self.jsd,self.distr)
        time5=time.time()
        print('weight',k,'differential substitutions complete in time', hrs_min_sec(time5-time4))

        print('weight',k,'normalization complete in time',hrs_min_sec(time.time()-time0))
        return (self.FN,self.KN)
    
    def der_term(self,F,K,i0,i1,i2,i3):
        """Returns the derivative term for the component of the Bianchi identity
        applied to (i0,i1,i2), which is an element of self.alg."""
        g=self.alg
        X0,X1,X2=[g.basis[a] for a in [i0,i1,i2]]
        w0,w1,w2=[None,None,None]
        try: w0=g.ext_alg.elt_from_cd({(str(X1),str(X2)):1})
        except(KeyError): pass
        try: w1=g.ext_alg.elt_from_cd({(str(X2),str(X0)):1})
        except(KeyError): pass
        try: w2=g.ext_alg.elt_from_cd({(str(X0),str(X1)):1})
        except(KeyError): pass

        r=0
        if w0!=None: r+=self.dir_der(K.apply_cochain_map(w0).vec[i3],F.col(i0))
        if w1!=None: r+=self.dir_der(K.apply_cochain_map(w1).vec[i3],F.col(i1))
        if w2!=None: r+=self.dir_der(K.apply_cochain_map(w2).vec[i3],F.col(i2))
        return -r
    
    def cb_term(self,F,K,i0,i1,i2,i3):
        """returns the coboundary of K applied to i0,i1,i2 columns of F.
        This is needed because we care about the value as a cochain in C(g,g), not just C(m,g)
        (at least for the purpose of checks)"""
        r=self.alg.elt()
        X0,X1,X2=[self.alg.basis[a] for a in [i0,i1,i2]]
    
        for i in range(3):
            Y0,Y1,Y2=[X0,X1,X2,X0,X1][i:i+3]
            w0=Y0.negative_projection().cast_as_ext_elt().wedge(Y1.negative_projection().cast_as_ext_elt())
            r+=K.apply_cochain_map(w0).ad(Y2)
            w1=Y0.ad(Y1).negative_projection().cast_as_ext_elt().wedge(Y2.negative_projection().cast_as_ext_elt())
            r+=K.apply_cochain_map(w1)
        return r.vec[i3]

    def gerst_term(self,F,K,i0,i1,i2,i3):
        X0,X1,X2=[self.alg.elt(F.col(a)) for a in [i0,i1,i2]]
        r=self.alg.elt()
        for i in range(3):
            Y0,Y1,Y2=[X0,X1,X2,X0,X1][i:i+3]
            print('Y0 =',Y0)
            print('Y1 =',Y1)
            print('Y2 =',Y2)
            Y0=Y0.negative_projection().cast_as_ext_elt()
            Y1=Y1.negative_projection().cast_as_ext_elt()
            Y2=Y2.negative_projection().cast_as_ext_elt()
            print('K.apply_cochain_map(Y0.wedge(Y1)) =',K.apply_cochain_map(Y0.wedge(Y1)))
            w=K.apply_cochain_map(Y0.wedge(Y1)).negative_projection().cast_as_ext_elt()
            print('w =',w)
            r+=K.apply_cochain_map(w.wedge(Y2))
        return r.vec[i3]
    
    def Bianchi(self,F,K,i0,i1,i2,i3,subdivide=False):
        """Returns the Bianchi expression corresponding to indices (i,j,k,l) in the frame F. 
        Should be zero modulo the Jacobi identity for the downstairs distribution"""
        r0=self.cb_term(F,K,i0,i1,i2,i3)
        r1=self.der_term(F,K,i0,i1,i2,i3)
        r2=self.gerst_term(F,K,i0,i1,i2,i3)
        if subdivide: return (r0, r1, r2)
        return D.normalize_der(r0+r1+r2)
```

## Prenormalization

From the involutivity conditions $[V_i,V_i]\subseteq V_i$ and $[\mathcal{J}^{(i)},V_i]\subseteq \mathcal{J}^{(i)}$, we can impose the following conditions on the structure function $K_{ij}^k$:

$$ [\varepsilon_i,\varepsilon_j]=\sum_{k=1}^j K_{ij}^k\varepsilon_k,\quad \text{for}\ i<j\leq m$$
$$ [\varepsilon_i,\varepsilon_j]=K_{ij}^XX+\sum_{k=1}^j K_{ij}^k\varepsilon_k,\quad \text{for}\ j>m\ \text{and}\ i+j<2m+1$$
All this is captured in the below "prenormalization", but the indices are shifted to reflect the index of each element in the basis $(Y,H,E,X,\varepsilon_1,\ldots \varepsilon_6,\eta)$.
