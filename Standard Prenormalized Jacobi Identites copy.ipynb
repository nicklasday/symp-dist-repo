```python
def add_subs_der(key,dir,subs_dict,distr):
    '''Adds and back substitutes 
    abn_ind_der(key,dir)-->abn_ind_der(subs_dict[key],dir)'''
    val=distr.normalize_der(distr.abn_ind_der(subs_dict[key],dir))
    subs_dict[distr.abn_ind_der(key,dir)]=val
    # Should I also substitute for the normalized version here?
    back_substitute(subs_dict,distr.abn_ind_der(key,dir),val)
```

```python
def wght_of_ind(ind_obj,alg):
    """Returns the weight of ind_obj with respect to the basis of alg
    Note: this is negative the weight of the corresponding tensor"""
    r=sum([alg.basis[i].wght for i in ind_obj.indices])
    return r-2*alg.basis[2].wght
```
