```python
def epsilon_proj(g_elt):
    temp=g_elt.vec
    for i in [0,1,2,3,10]:
        temp[i]=0
    return g.elt(temp)

def cochain_mat_rep(f):
    """Assuming f is a 1-cochain in C^1(m,g)"""
    r=sp.zeros(len(g.basis))
    for i in range(len(g.basis)-len(g.m_basis),len(g.basis)):
        v=f.apply_cochain_map(g.basis[i].cast_as_ext_elt()).vec
        for j in range(len(g.basis)):
            r[j,i]=v[j]
    return r

def one_cochain_from_mat_rep(M):
    """Only considers the m-->g part of the sympy matrix M"""
    r=C.elt({})
    for i in range(sp.shape(M)[0]):
        for j in range(len(g.basis)-len(g.m_basis),sp.shape(M)[1]):
            if M[i,j]!=0:
                r+=C.elt_from_cd({(g.basis_strs[j],g.basis_strs[i]):M[i,j]})
    return r

def neg_proj(g_elt):
    return g.elt([0,0,0]+g_elt.vec[3:len(g_elt.vec)])

# This is sufficiently useful to include in linear algebra helpers
def unipotent_inv(M,k=None):
    if k is None:
        A=M-sp.eye(sp.shape(M)[0])
        i=1
        Ai=A
        while Ai!=sp.zeros(sp.shape(M)[0]):
            Ai=Ai*A
            i+=1
    else: 
        i=k
    M_pwrs=[sp.eye(sp.shape(M)[0])]
    for j in range(1,i):
        M_pwrs.append(M*M_pwrs[-1])
    
    return sum([(-1)**(j+1)*sp.binomial(i,j)*M_pwrs[j-1] for j in range(1,i+1)],start=sp.zeros(sp.shape(M)[0]))
```
