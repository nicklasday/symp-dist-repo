```python
g=ts.SympSymb(9)
C=cc.CochainComplex(g)
```

```python
# Check that im(d^*) is g^0-invariant
for w in sorted(C.basis(2).keys()):
    if w>0:
        M=C.subspace_basis('coexact',2,w)
        N=C.subspace_basis('coexact',2,w+1)
        for j in range(sp.shape(M)[1]):
            c=C.elt({2:{w:M.col(j)}})
            v=(g.basis[0].ad(c)).vd[2][w+1]
            N1=N.col_insert(0,v)
            if N1.rank()!=N.rank(): 
                print('Failure:', c)
```
