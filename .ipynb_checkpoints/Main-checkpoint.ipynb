```python
def K_s(*ind):
    return K[*ind].subs(antisymm_K_subs).subs(Ps.K_subs)
```

```python
## Helpers for efficient application of the Jacobi identity
def max_nz_ind(i,j):
    """given indices i, j, returns the maximal index
    k such that K_{ij}^k is (possibly) nonzero
    """
    wi=g.basis[i].wght
    wj=g.basis[j].wght
    if wi+wj>1: return -1
    if wi+wj<=g.basis[-1].wght: return len(g.basis)-1
    k=0
    while g.basis[k].wght>=wi+wj:k+=1
    return k-1

def QNF_K(*ind,normalization='standard'):
    """returns K[*ind], possibly transposing antisymmetric indices,
    substituting using the given normalization
    INPUTS:
    * '*ind' -- at least 3 indices 
    * 'normalization' -- among 'standard', 'invol', and 'none' 
    """
    P=[Ps,Pi,Pn][['standard','invol','none'].index(normalization)]
    if ind[0]==ind[1]: return 0
    s=1
    t=list(ind)
    if ind[0]>ind[1]:
        s=-1
        t[0]=ind[1]
        t[1]=ind[0]
    if t[2]>max_nz_ind(t[0],t[1]): return 0
    if K[t[0],t[1],t[2]] in P.K_subs:
        return s*K[t[0],t[1],t[2]].subs(P.K_subs)
    return s*K[t[0],t[1],t[2]]
    
```

```python
JR={}
def compute_some_JR():
    time0=time.time()
    for i in range(len(g.basis)):
        for j in range(i+1,len(g.basis)):
            for k in range(j+1,len(g.basis)):
                for l in range(len(g.basis)):
                    t=QNF_K(j,k,l,i)+sum([QNF_K(j,k,r)*QNF_K(i,r,l) for r in range(len(g.basis))])
                    t=t+QNF_K(k,i,l,j)+sum([QNF_K(k,i,r)*QNF_K(j,r,l) for r in range(len(g.basis))])
                    t=t+QNF_K(i,j,l,k)+sum([QNF_K(i,j,r)*QNF_K(k,r,l) for r in range(len(g.basis))])
                    JR[(i,j,k,l)]=t
                    if time.time()-time0>60:
                        print('Final:',(i,j,k,l))
                        return None
```

```python
compute_some_JR()
```
