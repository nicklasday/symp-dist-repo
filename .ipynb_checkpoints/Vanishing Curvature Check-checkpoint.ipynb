```python
g=Symp_symb(7)
E=g.ext_alg
C=g.cochain_complex
P=Geom_Prolongation(g,prenorm='none')
# Pi=Geom_Prolongation(g,prenorm='invol')
# Ps=Geom_Prolongation(g,prenorm='standard')
y,h,e=symbols('y,h,e')
yn,hn,en=symbols('yn,hn,en')
K=IndexedBase('K')
Y,H,E,X,e1,e2,e3,e4,e5,e6,N=g.basis
p1_subs_dict={e:e+en,h:h+hn,y:exp(2*hn)*y+yn}
p1_inv=-exp(-2*hn)*yn*Y-hn*H-en*E
```

```python
def check_equivar(c):
    r=c.subs(p1_subs_dict)-p1_inv.Ad(c)
    simplify_cochain(r)
    if r==C.elt({}): return True
    else: return r
```

```python
def back_substitute(d):
    """Makes sure the keys of d don't show up in the values of d.
    This might not be airtight, but it's pretty good, I think."""
    effective=False
    complete_keys=set()
    while complete_keys!=set(d.keys()):
        for A in d:
            t=d[A].subs(d)
            if simplify(t)==simplify(d[A]): 
                complete_keys.add(A)
            else: 
                effective=True
                d[A]=t
    if effective: print('Back substitutions performed')
```
