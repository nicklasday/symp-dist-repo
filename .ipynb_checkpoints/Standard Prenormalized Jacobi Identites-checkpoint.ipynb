## Prenormalized Jacobi Identities for n=6 through weight 3

```python
def wght_of_ind(ind_obj,alg):
    """Returns the weight of ind_obj with respect to the basis of alg
    Note: this is negative the weight of the corresponding tensor"""
    r=sum([alg.basis[i].wght for i in ind_obj.indices])
    return r-2*alg.basis[ind_obj.indices[2]].wght
```

```python
def Add_JIS(K_list,JI_ind_list):
    for i in range(len(K_list)):
        s=solve(ds_subs(D.Jacobi_id(*JI_ind_list[i]),J_subs,D),K_list[i])
        if len(s)==0: print(JI_ind_list[i], 'has no solutions for ',K_list[i])
        ds_add_key(K_list[i],s[0],J_subs,D)
```

```python
J_subs={}
# Start with the prenormalization
for k in range(4,11-2):
    for k1 in range(3,k+1):
        J_subs[K[3,k,k1]]={tuple():0}
J_subs[K[3,9,9]]={tuple():0}
J_subs[K[3,9,8]]={tuple():0}

JI_ind_wghts={}
for i in range(3,len(g.basis)):
    for j in range(i+1,len(g.basis)):
        for k in range(j+1,len(g.basis)):
            for l in range(3,len(g.basis)):
                w=-g.basis[i].wght-g.basis[j].wght-g.basis[k].wght+g.basis[l].wght
                if not w in JI_ind_wghts: JI_ind_wghts[w]=[]
                JI_ind_wghts[w].append((i,j,k,l))

# wght 1
K_list=[K[4,5,5],K[4,6,6],K[4,7,7],K[4,8,8],K[3,10,10],K[5,6,7],K[5,7,8],K[6,8,10],K[5,9,10],K[6,7,9]]
JI_ind_list=[A for A in JI_ind_wghts[1] if A not in [(4,5,6,8),(4,5,7,9),(4,6,7,10)]]
Add_JIS(K_list,JI_ind_list)

# wght 2
K_list=[K[4,5,4],K[4,6,5],K[4,7,6],K[4,8,7],K[5,9,9],K[5,10,10],K[5,7,7],K[5,8,8],K[5,6,6],K[6,9,10],K[6,8,9],
        K[7,8,10],K[4,5,3],K[4,10,9],K[6,7,8]]
JI_ind_list=[A for A in JI_ind_wghts[2] if A not in [(4,5,7,8),(4,6,8,10),(4,6,7,9),(5,6,7,10)]]
Add_JIS(K_list,JI_ind_list)

# wght 3
K_list=[K[4,6,3],K[4,6,4],K[4,7,5],K[4,8,6],K[5,6,5],K[5,9,8],K[5,10,9],K[5,8,7],K[5,7,6],K[6,7,7],K[6,9,9],
        K[6,10,10],K[6,8,8],K[7,8,9],K[7,9,10],K[3,9,7],K[3,10,9,4],K[4,10,8],K[4,10,10,4,3]]
JI_ind_list=[A for A in JI_ind_wghts[3] if A not in [(4,5,7,7),(4,6,7,8),(4,5,10,10),
                                                 (4,6,8,9),(4,6,9,10),(4,7,8,10),(5,6,7,9),(5,6,8,10)]]
Add_JIS(K_list,JI_ind_list)

# wght 4
K_list=[K[5,6,3],K[5,6,4],K[5,7,5],K[5,8,6],K[5,9,7],K[4,10,7],K[4,7,4],K[6,7,6],K[4,8,5],K[6,9,8],K[6,10,9],
        K[6,8,7],K[7,8,8],K[7,9,9],K[7,10,10],K[3,10,7],K[8,9,10],K[3,10,8,4],K[5,8,9,4,3,3],K[4,7,3],K[5,8,9,4,4,3]]
JI_ind_list=[A for A in JI_ind_wghts[4] if A not in [(4,5,7,6),(4,6,7,7),(4,6,8,8),(4,6,9,9),(4,6,10,10),(4,7,8,9),
                                                     (4,7,9,10),(5,6,7,8),(5,6,8,9),(5,6,9,10),(5,7,8,10)]]
Add_JIS(K_list,JI_ind_list)

# wght 5
K_list=[K[5,7,3],K[5,7,4],K[5,8,5],K[5,9,6],K[5,10,7],K[4,8,3],K[4,8,4],K[4,9,5],K[6,8,6],K[6,9,7],
        K[6,10,8],K[6,7,5],K[7,8,7],K[7,9,8],K[7,10,9],K[4,9,7,3,3],K[8,9,9],K[8,10,10],
        K[3,9,5],K[5,10,8,4],K[5,10,8,3],K[4,10,6],K[4,9,9,3,4,3,3],K[5,8,9,4,3,4,3],K[5,8,9,3,4,4,3]]
JI_ind_list=[A for A in JI_ind_wghts[5] if A not in [(4,5,7,5),(4,6,7,6),(4,6,8,7),(4,6,9,8),(4,7,8,8),(4,7,9,9),
                                                     (4,7,10,10),(4,8,9,10),(5,6,7,7),(5,6,8,8),(5,6,9,9),(5,6,10,10),
                                                     (5,7,8,9),(5,7,9,10),(6,7,8,10)]]
Add_JIS(K_list,JI_ind_list)

# wght 6
K_list=[K[5,8,3],K[4,9,4],K[5,9,5],K[5,10,6],K[6,7,3],K[6,7,4],K[6,8,5],K[6,9,6],K[6,10,7],K[5,8,4],K[7,8,6],K[7,9,7],
        K[7,10,8],K[4,9,8,3,3,3,3],K[8,9,8],K[8,10,9],K[9,10,10],K[4,9,3],K[5,8,9,3,3,3,3,4],K[4,10,10,3,4,3,3,4],
        K[3,10,8,3,4,4],K[5,8,9,3,4,3,3,4],K[3,10,9,3,4,3,4]]
JI_ind_list=[A for A in JI_ind_wghts[6] if A not in [(3,8,9,9),(4,5,7,3),(4,5,7,4),(4,6,7,5),(4,6,8,6),
                                                     (4,6,9,7),(4,7,8,7),(4,7,9,8),(4,8,9,9),(4,8,10,10),
                                                     (5,6,7,6),(5,6,8,7),(5,6,9,8),(5,7,8,8),(5,7,9,9),
                                                     (5,7,10,10),(5,8,9,10),(6,7,8,9),(6,7,9,10)]]
Add_JIS(K_list,JI_ind_list)

# wght 7
K_list=[K[5,9,3],K[5,9,4],K[5,10,5],K[6,8,3],K[6,8,4],K[6,9,5],K[6,10,6],K[4,10,5,3],K[3,10,4],K[7,8,5],
        K[7,9,6],K[7,10,7],K[4,10,10,3,3,3,3,3,3],K[8,9,7],K[8,10,8],K[4,9,9,3,3,3,3,3,3],K[9,10,9],K[3,10,6,4,4],
        K[3,10,5,4],K[3,9,6,4,4,3],K[4,9,6,3,4,4],K[3,10,8,3,3,4,4],K[5,8,9,3,4,3,4,3,3],K[5,8,9,3,3,4,3,4,3],
        K[5,8,9,3,4,3,3,3,4],K[3,10,9,3,3,3,4,4]]
JI_ind_list=[A for A in JI_ind_wghts[7] if A not in [(3,9,10,10),(4,6,7,3),(4,6,7,4),(4,6,8,5),(4,6,9,6),(4,7,8,6),
                                                     (4,7,9,7),(4,8,9,8),(4,9,10,10),(5,6,7,5),(5,6,8,6),(5,6,9,7),
                                                     (5,7,8,7),(5,7,9,8),(5,8,9,9),(5,8,10,10),(6,7,8,8),(6,7,9,9),
                                                     (6,7,10,10),(6,8,9,10)]]
Add_JIS(K_list,JI_ind_list)

# wght 8
K_list=[]
JI_ind_list=[A for A in JI_ind_wghts[8] if A not in []]
Add_JIS(K_list,JI_ind_list)

# wght 9
K_list=[]
JI_ind_list=[A for A in JI_ind_wghts[9] if A not in []]
Add_JIS(K_list,JI_ind_list)
```
