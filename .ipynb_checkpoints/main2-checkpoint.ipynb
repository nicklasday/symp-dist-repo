```python
def max_nz_ind(i,j):
    w=g.basis[i].wght+g.basis[j].wght
    print('w =',w)
    if w>1: return -1
    if w<=g.basis[-1].wght: return len(g.basis)-1
    k=0
    while k<len(g.basis) and g.basis[k].wght>=w:
        # print('since g.basis',[k],'.wght =',g.basis[k].wght)
        k+=1
    return k-1
```

```python
print(max_nz_ind(3,4)==5)
print(max_nz_ind(0,0)==-1)
print(max_nz_ind(1,2)==2)
print(max_nz_ind(4,5)==6)
print(max_nz_ind(5,6)==8)
print(max_nz_ind(5,9)==10)
```
