```python
g=Symp_symb(7)
E=g.ext_alg
C=g.cochain_complex
P=Geom_Prolongation(g,prenorm='none')
# Pi=Geom_Prolongation(g,prenorm='invol')
# Ps=Geom_Prolongation(g,prenorm='standard')
y,h,e=symbols('y,h,e')
K=IndexedBase('K')
Y,H,E,X,e1,e2,e3,e4,e5,e6,N=g.basis
```

```python
yn,hn,en=symbols('yn,hn,en')
p1_subs_dict={e:e+en,h:h+hn,y:exp(2*hn)*y+yn}
p1_inv=-exp(-2*hn)*yn*Y-hn*H-en*E

def check_equivar(c):
    r=c.subs(p1_subs_dict)-p1_inv.Ad(c)
    simplify_cochain(r)
    if r==C.elt({}): return True
    else: return r
```

## Jacobi Identities for n=6 through weight 2

```python
wght_1_list=[]
for i in range(3,len(g.basis)):
    for j in range(i+1,len(g.basis)):
        for k in range(j+1,len(g.basis)):
            for l in range(3,len(g.basis)):
                if g.basis[i].wght+g.basis[j].wght+g.basis[k].wght-g.basis[l].wght==-1:
                    wght_1_list.append((i,j,k,l))
```

```python
J_subs={}
J_subs[1]={}
K_list=[K[4,5,5],K[4,6,6],K[4,7,7],K[4,8,8],K[3,10,10],K[5,6,7],K[5,7,8],K[3,5,5],K[3,6,6],K[5,8,9]]
JI_ind_list=[A for A in wght_1_list if A not in [(4,5,6,8),(4,5,7,9),(4,6,7,10)]]
for i in range(len(K_list)):
    print('\n\n',P.Jacobi_id(*JI_ind_list[i]).subs(J_subs[1]))
    print('has solutions for ',K_list[i])
    s=solve(P.Jacobi_id(*JI_ind_list[i]).subs(J_subs[1]),K_list[i])
    print(s)
    J_subs[1][K_list[i]]=solve(P.Jacobi_id(*JI_ind_list[i]).subs(J_subs[1]),K_list[i])[0]
```

```python
for A in J_subs[1]:
    J_subs[1][A]=J_subs[1][A].subs(J_subs[1])
    print(A,'-->',J_subs[1][A])
```

```python
for A in wght_1_list:
    if P.Jacobi_id(*A).subs(J_subs[1])!=0: print(A)
print('Jacobi Identity weight 1 test complete')
```

```python
wght_2_list=[]
for i in range(3,len(g.basis)):
    for j in range(i+1,len(g.basis)):
        for k in range(j+1,len(g.basis)):
            for l in range(3,len(g.basis)):
                if g.basis[i].wght+g.basis[j].wght+g.basis[k].wght-g.basis[l].wght==-2:
                    wght_2_list.append((i,j,k,l))
```

```python
J_subs[2]={}
# Add in derivatives
for A in list(J_subs[1].keys()):
    J_subs[2][P.abn_ind_der(A,3)]=P.abn_ind_der(J_subs[1][A],3)
    J_subs[2][P.abn_ind_der(A,4)]=P.abn_ind_der(J_subs[1][A],4)
K_list=[K[3,5,3],K[4,6,5],K[4,7,6],K[4,8,7],K[4,9,8],K[5,10,10],K[5,7,7],K[5,8,8],K[5,9,9],K[6, 9, 10],
       K[6, 8, 9],K[7, 8, 10],K[4, 5, 3],0,K[4, 10, 9],K[6, 7, 8],0]
JI_ind_list=[A for A in wght_2_list if A not in []]
for i in range(len(K_list)):
    if K_list[i]!=0:
        t=P.Jacobi_id(*JI_ind_list[i]).subs(J_subs[1]).subs(J_subs[2])
        print('\n\n',JI_ind_list[i])
        print(t)
        print('has solutions for ',K_list[i])
        s=solve(t,K_list[i])
        print(s)
        J_subs[2][K_list[i]]=s[0]
```

```python
# Make sure substitutions are complete
complete_2=set()
for A in J_subs[2]:
    temp=J_subs[2][A].subs(J_subs[1])
    if J_subs[2][A]==temp:complete_2.add(A)
    else: J_subs[2][A]=temp
if complete_2==set(J_subs[2].keys()): print('Substitutions complete')
else: print('More substitution needed')
```

```python
# Make sure substitutions are complete
complete_2=set()
for A in J_subs[2]:
    temp=J_subs[2][A].subs(J_subs[2])
    if J_subs[2][A]==temp:complete_2.add(A)
    else: J_subs[2][A]=temp
if complete_2==set(J_subs[2].keys()): print('Substitutions complete')
else: print('More substitution needed')
```

```python
# Make sure substitutions are complete
complete_1=set()
for A in J_subs[1]:
    temp=J_subs[1][A].subs(J_subs[1]).subs(J_subs[2])
    if J_subs[1][A]==temp:complete_1.add(A)
    else: J_subs[1][A]=temp
if complete_1==set(J_subs[1].keys()): print('Substitutions complete')
else: print('More substitution needed')
```

```python
total_subs=copy.copy(J_subs[1])
total_subs.update(J_subs[2])
```

```python
for A in wght_2_list:
    if simplify(P.Jacobi_id(*A).subs(total_subs))!=0: print(A)
print('Jacobi Identity weight 2 test complete')
```
