```python
from sympy import *
import time
from collections import defaultdict
```

```python
# We'll work with gl2+heis(2m+1)
m=3
```

```python
# i=Idx('i',range=(1,2*m+1))
# j=Idx('j',range=(1,2*m+1))
# k=Idx('k',range=(1,2*m+1))
c=IndexedBase('c',shape=(2*m,2*m,2*m))
f=Index
Y,H,E,X=symbols('Y,H,E,X')
substitutions={'antisymm':{},Y:{},H:{},E:{},X:{}}
total_subs={}
determined_cijk=set()
rels={Y:[0],H:[0],E:[0],X:[0]}

# Set the antisymmetric relations as substitutions
for a in range(1,2*m+1):
    for d in range(1,2*m+1):
        substitutions['antisymm'][c[a,a,d]]=0
        for b in range(a+1,2*m+1):
            substitutions['antisymm'][c[b,a,d]]=-c[a,b,d]
total_subs.update(substitutions['antisymm'])
```

```python
for i in range(1,2*m+1):
    rels[Y].append(sum([(b-1)*(2*m+1-b)*c[i,b,b-1] for b in range(2,2*m+1)]).subs(substitutions['antisymm']))
    rels[H].append(sum([(2*b-2*m-1)*c[i,b,b] for b in range(1,2*m+1)]).subs(substitutions['antisymm']))
    rels[E].append(sum([c[i,b,b] for b in range(1,2*m+1)]).subs(substitutions['antisymm']))
    rels[X].append(sum([c[i,b,b+1] for b in range(1,2*m)]).subs(substitutions['antisymm']))
```

```python
def c_index(c1):
    '''c1: An Idx between 0 and 2*m+1
       returns: the unraveled index of c1 in c'''
    ctr=0
    while c1!=c[ctr//((2*m+1)**2),(ctr//(2*m+1))%(2*m+1),ctr%(2*m+1)]:
        ctr+=1
    return ctr
```

```python
def update_subs_dicts(subs_dict,gl_elt):
    '''subs_dict: A dict representing a cijk substitution
       result: None
       updates the dicts substitutions and total_subs by subs_dict'''
    for A in [Y,H,E,X]:
        for key in substitutions[A]:
            # 'Back substitute'
            if hasattr(substitutions[A][key],'subs'):
                substitutions[A][key]=substitutions[A][key].subs(subs_dict)
            # Add the substitution to the dict gl_elt
        if A==gl_elt: substitutions[A].update(subs_dict)
    for key in total_subs:
        if hasattr(total_subs[key],'subs'):
            total_subs[key]=total_subs[key].subs(subs_dict)
    total_subs.update(subs_dict)
    for key in subs_dict:
        determined_cijk.add(key)
```

```python
# Let's use the relations to construct substitution dictionaries.
# The determined cijk will be kept as the keys of total_subs

for A in [E,X,H,Y]:
    for r in range(1,len(rels[A])):
        curr_list=list(rels[A][r].as_coefficients_dict().keys())
        curr_list.sort(key=c_index)
        ctr=0
        while True:
            # Find the least cijk in curr_list which is not yet determined
            # Note that rels[A][r] contains only cijk with i<j
            if not (curr_list[ctr] in total_subs):
                new_cijk=curr_list[ctr]
                
                # Add the relation to substitutions
                new_subs_dict={new_cijk:solve(rels[A][r].subs(total_subs),new_cijk)[0]}
                update_subs_dicts(new_subs_dict,A)
                break
            else: ctr+=1
```
